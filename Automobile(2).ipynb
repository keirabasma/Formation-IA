```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Charger le fichier CSV (en supposant qu'il est nommé 'Automobile.csv')
# Si votre fichier s'appelle 'malak', remplacez le nom du fichier.
df = pd.read_csv('Automobile.csv')

# --- Nettoyage des Données ---

# 1. Remplacer les chaînes '?' par NaN (valeurs manquantes)
df.replace('?', pd.NA, inplace=True)

# 2. Convertir 'horsepower' et 'price' en numérique
# 'coerce' va transformer les valeurs qui ne peuvent pas être converties (les NaN après le remplacement) en NaN.
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
df['price'] = pd.to_numeric(df['price'], errors='coerce')

# 3. Supprimer les lignes où 'horsepower' ou 'price' sont manquants (pour le tracé)
malak_clean = df.dropna(subset=['horsepower', 'price'])

# Note : J'ai renommé le DataFrame nettoyé en 'malak_clean' pour qu'il soit propre.
```

```python

df = pd.read_csv('Automobile.csv', na_values=['?'])
df['horsepower'] = pd.to_numeric(df['horsepower'])
df['price'] = pd.to_numeric(df['price'])

malak_clean = df.dropna(subset=['horsepower', 'price'])
print(f"Nombre de lignes prêtes pour le tracé : {len(malak_clean)}")
```

```python


g = sns.jointplot(
    x="horsepower", 
    y="price", 
    data=malak_clean,  # Utilisez le DataFrame nettoyé
    kind='hex', 
    height=8,        
    ratio=5,         
    cmap='Greens', 
    marginal_kws=dict(bins=20, color="darkgreen", alpha=0.5) 
)

g.ax_joint.set_xlabel("Horsepower")
g.ax_joint.set_ylabel("Price")

plt.show()
```
