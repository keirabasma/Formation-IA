1️⃣ First DataSet

Import libraries & load data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
files.upload()
df = pd.read_csv("cars.csv")
df.head()
```

Basic data handling (Pandas)

```python
df.info()
df.describe()
```

NumPy usage

```python
price_np = np.array(df["price"])
avg_price = np.mean(price_np)

print("Average price:", avg_price)
```

Simple visualizations (Matplotlib / Seaborn)


```python
#Price distribution
plt.hist(df["price"], bins=20)
plt.xlabel("Price")
plt.ylabel("Count")
plt.title("Car Price Distribution")
plt.show()

#Vehicle age vs price
sns.scatterplot(data=df, x="vehicle_age", y="price")
plt.title("Vehicle Age vs Price")
plt.show()

#Average price by fuel type
sns.barplot(data=df, x="fuel_type", y="price")
plt.title("Average Price by Fuel Type")
plt.show()
```

Same idea using Plotly (simple)

2️⃣ Secend DataSet

```python
# Dataset Name: Students Performance in Exams (Source: Kaggle)

# 1️⃣ Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from google.colab import files
uploaded = files.upload()  # select "StudentsPerformance.csv"

df = pd.read_csv("StudentsPerformance.csv")
df.head()

# 3️⃣ Basic data handling (Pandas)
df.info()
df.describe()

# 4️⃣ Explicit NumPy usage (required)
# Calculate the average Math score
math_np = np.array(df["math score"])
avg_math = np.mean(math_np)
print("Average Math Score:", avg_math)

# 5️⃣ Visualization with Matplotlib
# Histogram of Math Scores
plt.figure(figsize=(8, 5))
plt.hist(math_np, bins=20, color='skyblue', edgecolor='black')
plt.xlabel("Math Score")
plt.ylabel("Count")
plt.title("Math Score Distribution")
plt.show()

# 6️⃣ Visualization with Seaborn
# Scatter plot: Reading Score vs Math Score
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="reading score", y="math score")
plt.title("Reading Score vs Math Score")
plt.show()

# 7️⃣ Visualization with Plotly
# Interactive Scatter: Writing Score vs Math Score
fig = px.scatter(
    df,
    x="writing score",
    y="math score",
    title="Math Score vs Writing Score"
)
fig.show()
```
