## **Part 1 : Data Visualization on the PADOC Dataset (gapminder.csv)**

```python
import pandas as pd
```

```python
df = pd.read_csv("gapminder.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.drop(columns=["Unnamed: 0"], inplace=True)
```

```python
import matplotlib.pyplot as plt

plt.hist(df["life_exp"], bins=20)
plt.xlabel("Life Expectancy")
plt.ylabel("Frequency")
plt.title("Distribution of Life Expectancy")
plt.show()
```

```python
import seaborn as sns

sns.boxplot(x="continent", y="life_exp", data=df)
plt.title("Life Expectancy by Continent")
plt.show()
```

```python
plt.scatter(df["gdp_cap"], df["life_exp"])
plt.xlabel("GDP per Capita")
plt.ylabel("Life Expectancy")
plt.title("GDP per Capita vs Life Expectancy")
plt.show()
```

```python
sns.scatterplot(
    x="gdp_cap",
    y="life_exp",
    hue="continent",
    data=df
)
plt.title("GDP vs Life Expectancy by Continent")
plt.show()
```

```python
plt.scatter(df["population"], df["life_exp"])
plt.xlabel("Population")
plt.ylabel("Life Expectancy")
plt.title("Population vs Life Expectancy")
plt.show()
```

```python
corr = df[["year", "population", "life_exp", "gdp_cap"]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

# **Part 2 : Data Visualization on my Dataset (netflix_titles.csv)**

```python
import pandas as pd

df = pd.read_csv("netflix_titles.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.dropna(inplace=True)
```

```python
df.hist(figsize=(10,8))
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(data=df.select_dtypes(include="number"))
plt.show()
```

```python
sns.countplot(x=df.select_dtypes(include="object").columns[0], data=df)
plt.xticks(rotation=45)
plt.show()
```
