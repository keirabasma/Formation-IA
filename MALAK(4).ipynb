```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
df = pd.read_csv("cars.csv")
df.head() 
malak = df
malak.head() 
```

```python
malak["price"].value_counts()
```

```python
malak.plot(kind="scatter", x="vehicle_age", y="km_driven")
plt.show()
```

```python
malak.plot(kind="scatter", x="vehicle_age", y="km_driven",color="green",s=70)
plt.show() 	
```

```python
sns.jointplot(x="vehicle_age", y="km_driven", data=malak, size=5)
plt.show()
```

```python
sns.FacetGrid(malak, hue="model", height=3, aspect=3) \
   .map(plt.scatter, "vehicle_age", "km_driven") \
   .add_legend()
plt.show()
```

```python
sns.boxplot(x="model", y="vehicle_age", data=malak )
plt.show()
```

```python
ax= sns.boxplot(x="modle", y="vehicle_age", data=malak)
ax= sns.stripplot(x="modle", y="vehicle_age", data=malak, jitter=True, edgecolor="gray")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6)) 
ax = sns.boxplot(x="price", y="km_driven", data=malak)
ax = sns.stripplot(x="price", y="km_driven", data=malak, jitter=True, edgecolor="gray", color=".3") 
if len(ax.patches) > 6:
    boxtwo = ax.patches[6]
    boxtwo.set_facecolor('red')
    boxtwo.set_edgecolor('black')
if len(ax.patches) > 4:
    boxthree = ax.patches[4]
    boxthree.set_facecolor('yellow')
    boxthree.set_edgecolor('black')
plt.title('Diagramme en boîte et à bandes de km_driven par price') 
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
sns.violinplot(x="price", y="vehicle_age", data=malak) 
plt.show()
```
