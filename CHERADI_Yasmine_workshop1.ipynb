```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/gapminder.csv")
print(df.head())
```

```python
print(df.head())
print(df.describe())
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['life_exp'], bins=8, kde=True)
plt.title("Distribution de l'espérance de vie")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='country', y='life_exp', data=df)
plt.xticks(rotation=45)
plt.title("Espérance de vie par pays")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='continent', y='life_exp', data=df)
plt.title("Espérance de vie par continent")
plt.show()
```
