**QUESTION 2 :** utilisation de plotly pour le dataset "**gapminder.csv**".

```python
import pandas as pd

# Load CSV
df = pd.read_csv("gapminder.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
# Check for missing values
print(df.isnull().sum())

# Check unique continents
print(df['continent'].unique())

# Convert data types if necessary
df['year'] = df['year'].astype(int)
df['population'] = df['population'].astype(int)
df['life_exp'] = df['life_exp'].astype(float)
df['gdp_cap'] = df['gdp_cap'].astype(float)
```

```python
import plotly.express as px

fig = px.scatter(
    df,
    x="gdp_cap",
    y="life_exp",
    size="population",
    color="continent",
    hover_name="country",
    title="GDP per Capita vs Life Expectancy (2007)",
    log_x=True,  # log scale for better spread
    size_max=60
)
fig.show()
```

```python
fig = px.bar(
    df.sort_values("population", ascending=False),
    x="country",
    y="population",
    color="continent",
    title="Population by Country (2007)"
)
fig.show()
```

```python
fig = px.box(
    df,
    x="continent",
    y="life_exp",
    color="continent",
    title="Life Expectancy Distribution by Continent"
)
fig.show()
```

```python
fig = px.choropleth(
    df,
    locations="country",
    locationmode="country names",
    color="gdp_cap",
    hover_name="country",
    color_continuous_scale=px.colors.sequential.Plasma,
    title="GDP per Capita by Country (2007)"
)
fig.show()
```

```python
fig = px.pie(
    df.groupby("continent")["population"].sum().reset_index(),
    values="population",
    names="continent",
    title="Population Share by Continent"
)
fig.show()
```

```python
fig = px.histogram(
    df,
    x="life_exp",
    nbins=20,
    color="continent",
    title="Life Expectancy Distribution by Continent",
    marginal="box"  # add box plot on top
)
fig.show()
```

**INTERPRÉTATIONS DES RÉSULTATS VISUALISÉS :**
Les visualisations montrent une forte corrélation positive entre le PIB par habitant et l’espérance de vie : les pays riches, principalement européens, tendent à avoir une espérance de vie élevée, tandis que les pays à faible PIB par habitant, comme l’Afghanistan ou l’Angola, présentent une espérance de vie plus basse. La taille de la population influence également les représentations visuelles, avec des pays comme l’Algérie apparaissant en position intermédiaire malgré un PIB moyen. Les analyses par continent révèlent des disparités régionales : l’Europe bénéficie d’une espérance de vie élevée et stable, l’Afrique montre une grande variabilité et l’Asie se situe dans des valeurs intermédiaires ou basses. Les cartes choroplèthes et les graphiques sectoriels soulignent l’inégalité géographique du développement économique et la répartition de la population, tandis que les histogrammes confirment la concentration des pays entre 40 et 75 ans d’espérance de vie. Globalement, ces données mettent en évidence que les pays pauvres avec une population importante font face à des défis majeurs en matière de santé, d’éducation et de services sociaux, et que les outils visuels sont efficaces pour identifier rapidement les tendances et les inégalités.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("/content/ground_water_quality_2018.csv")
print(df.head())
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(' ', '_')
    .str.replace('.', '', regex=False)
    .str.replace('/', '_')
)

df.columns
```

```python
plt.figure()
plt.hist(df['pH'], bins=15)
plt.xlabel("pH")
plt.ylabel("Frequency")
plt.title("Distribution of pH in Groundwater")
plt.show()
```

```python
plt.figure()
sns.boxplot(
    data=df,
    x='Classification',
    y='TDS'
)
plt.title("TDS Distribution by Water Classification")
plt.show()
```

```python
plt.figure()
plt.scatter(df['EC'], df['TDS'])
plt.xlabel("Electrical Conductivity (EC)")
plt.ylabel("Total Dissolved Solids (TDS)")
plt.title("EC vs TDS")
plt.show()
```

```python
plt.figure()
sns.scatterplot(
    data=df,
    x='EC',
    y='SAR',
    hue='Classification'
)
plt.title("EC vs SAR by Classification")
plt.show()
```

```python
sns.pairplot(
    df,
    vars=['pH','EC','TDS','Na','Cl','SO4'],
    hue='Classification'
)
plt.show()
```

```python
fig = px.scatter(
    df,
    x='EC',
    y='TDS',
    color='Classification',
    size='SAR',
    hover_data=['village','pH','NO3'],
    title="EC vs TDS (Interactive)"
)
fig.show()
```

```python
fig = px.box(
    df,
    y='NO3',
    color='Classification',
    points='all',
    title="Nitrate (NO3) Concentration by Classification"
)
fig.show()
```

```python
ions = ['Na','K','Ca','Mg','Cl','SO4']

fig = go.Figure()

for _, row in df.iterrows():
    fig.add_trace(go.Scatterpolar(
        r=[row[i] for i in ions],
        theta=ions,
        fill='toself',
        name=row['village']
    ))

fig.update_layout(
    polar=dict(radialaxis=dict(visible=True)),
    title="Radar Chart of Major Ions Composition"
)

fig.show()
```

**INTERPRÉTATIONS DES RÉSULTATS VISUALISÉS :** L’analyse montre que la qualité des eaux souterraines en 2018 dans les zones d’étude est généralement alcaline, avec une minéralisation principalement contrôlée par les sels dissous, comme l’indique la forte relation entre la conductivité électrique (CE) et les solides dissous totaux (TDS). Les classes de salinité élevées sont associées à des concentrations accrues en sodium, chlorure et sulfate, ce qui peut limiter l’aptitude à l’irrigation en raison des risques de salinité et de sodicité. La variabilité spatiale de la qualité des eaux souterraines suggère l’existence de processus hydrochimiques localisés, soulignant l’importance d’un suivi régulier et d’une gestion durable des ressources en eau souterraine.
