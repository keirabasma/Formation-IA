**QUESTION 2 :** utilisation de plotly pour le dataset "**gapminder.csv**".

```python
import pandas as pd

# Load CSV
df = pd.read_csv("gapminder.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
# Check for missing values
print(df.isnull().sum())

# Check unique continents
print(df['continent'].unique())

# Convert data types if necessary
df['year'] = df['year'].astype(int)
df['population'] = df['population'].astype(int)
df['life_exp'] = df['life_exp'].astype(float)
df['gdp_cap'] = df['gdp_cap'].astype(float)
```

```python
import plotly.express as px

fig = px.scatter(
    df,
    x="gdp_cap",
    y="life_exp",
    size="population",
    color="continent",
    hover_name="country",
    title="GDP per Capita vs Life Expectancy (2007)",
    log_x=True,  # log scale for better spread
    size_max=60
)
fig.show()
```

```python
fig = px.bar(
    df.sort_values("population", ascending=False),
    x="country",
    y="population",
    color="continent",
    title="Population by Country (2007)"
)
fig.show()
```

```python
fig = px.box(
    df,
    x="continent",
    y="life_exp",
    color="continent",
    title="Life Expectancy Distribution by Continent"
)
fig.show()
```

```python
fig = px.choropleth(
    df,
    locations="country",
    locationmode="country names",
    color="gdp_cap",
    hover_name="country",
    color_continuous_scale=px.colors.sequential.Plasma,
    title="GDP per Capita by Country (2007)"
)
fig.show()
```

```python
fig = px.pie(
    df.groupby("continent")["population"].sum().reset_index(),
    values="population",
    names="continent",
    title="Population Share by Continent"
)
fig.show()
```

```python
fig = px.histogram(
    df,
    x="life_exp",
    nbins=20,
    color="continent",
    title="Life Expectancy Distribution by Continent",
    marginal="box"  # add box plot on top
)
fig.show()
```

**INTERPRÉTATIONS DES RÉSULTATS VISUALISÉS :**
Les visualisations montrent une forte corrélation positive entre le PIB par habitant et l’espérance de vie : les pays riches, principalement européens, tendent à avoir une espérance de vie élevée, tandis que les pays à faible PIB par habitant, comme l’Afghanistan ou l’Angola, présentent une espérance de vie plus basse. La taille de la population influence également les représentations visuelles, avec des pays comme l’Algérie apparaissant en position intermédiaire malgré un PIB moyen. Les analyses par continent révèlent des disparités régionales : l’Europe bénéficie d’une espérance de vie élevée et stable, l’Afrique montre une grande variabilité et l’Asie se situe dans des valeurs intermédiaires ou basses. Les cartes choroplèthes et les graphiques sectoriels soulignent l’inégalité géographique du développement économique et la répartition de la population, tandis que les histogrammes confirment la concentration des pays entre 40 et 75 ans d’espérance de vie. Globalement, ces données mettent en évidence que les pays pauvres avec une population importante font face à des défis majeurs en matière de santé, d’éducation et de services sociaux, et que les outils visuels sont efficaces pour identifier rapidement les tendances et les inégalités.






**QUESTION 3 :** refaire la meme chose pour un autre dataset, dans mon cas, j'ai choisi de travailler avec le dataset "**cars.csv**"

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("cars.csv")  # Remplacez par le chemin vers votre fichier
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
def extract_numeric(value):
    try:
        return float(str(value).split()[0])
    except:
        return np.nan

df['mileage'] = df['mileage'].apply(extract_numeric)
df['engine'] = df['engine'].apply(extract_numeric)
df['max_power'] = df['max_power'].apply(extract_numeric)
df['seats'] = pd.to_numeric(df['seats'], errors='coerce')

# Supprimer les lignes avec valeurs manquantes pour simplifier
df.dropna(inplace=True)
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='km_driven', y='price', hue='fuel_type', size='vehicle_age', alpha=0.7)
plt.title('Price vs Km Driven')
plt.xlabel('Km Driven')
plt.ylabel('Price')
plt.show()
```

```python
plt.figure(figsize=(12,6))
brand_price = df.groupby('brand')['price'].mean().sort_values(ascending=False)
sns.barplot(x=brand_price.index, y=brand_price.values, palette='viridis')
plt.xticks(rotation=45)
plt.title('Average Price by Car Brand')
plt.ylabel('Average Price')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='transmission_type', y='price', palette='Set2')
plt.title('Price by Transmission Type')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['vehicle_age'], bins=15, kde=True, color='skyblue')
plt.title('Distribution of Vehicle Age')
plt.xlabel('Vehicle Age (years)')
plt.show()
```

```python
plt.figure(figsize=(10,7))
numeric_cols = ['vehicle_age', 'km_driven', 'mileage', 'engine', 'max_power', 'seats', 'price']
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```

```python
fig = px.scatter(df, x='km_driven', y='price', color='fuel_type', size='vehicle_age',
                 hover_data=['car_name','brand','model'], title='Price vs Km Driven (Plotly)')
fig.show()
```

```python
fig = px.box(df, x='fuel_type', y='price', color='fuel_type', title='Price by Fuel Type')
fig.show()
```

```python
fig = px.histogram(df, x='vehicle_age', nbins=15, color='fuel_type', title='Vehicle Age Distribution')
fig.show()
```

```python
brand_avg_price = df.groupby('brand')['price'].mean().reset_index().sort_values(by='price', ascending=False)
fig = px.bar(brand_avg_price, x='brand', y='price', color='price', title='Average Price per Brand')
fig.show()
```

```python
import plotly.figure_factory as ff
corr_matrix = df[numeric_cols].corr().values
fig = ff.create_annotated_heatmap(corr_matrix, x=numeric_cols, y=numeric_cols, colorscale='Viridis', showscale=True)
fig.update_layout(title='Correlation Heatmap (Plotly)')
fig.show()
```

**INTERPRÉTATIONS DES RÉSULTATS VISUALISÉS :** L’analyse exploratoire du dataset montre plusieurs tendances intéressantes. Tout d’abord, il existe une corrélation négative claire entre le kilométrage et le prix : plus un véhicule a roulé, plus sa valeur tend à être faible, avec des variations liées au type de carburant et à l’âge du véhicule. Les barplots révèlent que certaines marques affichent un prix moyen nettement plus élevé, ce qui reflète leur positionnement sur le marché ou leur réputation. Les boxplots mettent en évidence que les voitures à transmission automatique sont généralement plus chères et présentent une dispersion de prix plus large que celles à transmission manuelle. L’histogramme de l’âge des véhicules montre que la majorité des voitures ont entre 2 et 8 ans, tandis que la heatmap de corrélation indique que le prix est fortement influencé par des facteurs comme le moteur, la puissance et le nombre de sièges. Les visualisations interactives avec Plotly confirment ces tendances et permettent d’explorer plus facilement les relations entre plusieurs variables simultanément. Dans l’ensemble, ces résultats soulignent que le prix des véhicules dépend à la fois de leurs caractéristiques techniques et de facteurs liés à l’usure, offrant des insights utiles pour les acheteurs, les vendeurs et les analystes du marché automobile.
