- **Creating and Deleting Columns**

```python
df ['km_per_year'] = df['vehicle_age'] / df ['km_driven']

pd.reset_option('display.max_columns') # reset display option in default mode
print(df)
```

```python
df.drop ('km_per_year' , axis= 1 , inplace = True) 
# removes the column named 'km_per_year' from the DataFrame df by specifying 
# axis=1 to indicate a column operation, and inplace=True to apply the change 
# directly without reassigning the DataFrame.

print(df)
```

- **Display the row with the minimum price**

```python
df[df['price'] == df['price'].min()]
```

- **Sorting values**

```python
df.sort_values (by = 'price' , ascending = False)
```
