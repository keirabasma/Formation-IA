```python

import pandas as pd


df = pd.read_csv("cars.csv")


print("Columns:")
print(df.columns)
print("\nData types of columns:")
print(df.dtypes)
print("\nDataFrame shape (rows × columns):")
print(df.shape)


fuel_unique_count = df["fuel_type"].nunique()
print(f"\nNumber of unique values in 'fuel_type': {fuel_unique_count}")

fuel_counts = df["fuel_type"].value_counts()
print("\nCounts of each fuel type:")
print(fuel_counts)


df["price_per_km"] = df["price"] / df["km_driven"]

print("\nFirst 5 rows with the calculated column:")
print(df.head())


top_10_expensive = df.sort_values(by="price", ascending=False).head(10)

print("\nTop 10 most expensive cars:")
print(top_10_expensive)
```
