import numpy as np

import seaborn as sns
import plotly.express as px

print("Tout est prêt ")


### Analyse exploratoire
Ce dataset contient des mesures de fleurs Iris appartenant à trois espèces.
Les données sont numériques et bien équilibrées entre les classes.


```python
# Charger le dataset Iris
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

# Afficher les 5 premières lignes
df.head()
```

```python
import numpy as np

sepal_length = np.array(df['sepal_length'])

print("Moyenne :", np.mean(sepal_length))
print("Écart-type :", np.std(sepal_length))
print("Minimum :", np.min(sepal_length))
print("Maximum :", np.max(sepal_length))
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['sepal_length'], bins=20)
plt.title("Distribution de la longueur du sépale")
plt.xlabel("Sepal Length")
plt.ylabel("Fréquence")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['sepal_length'], df['petal_length'])
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Relation Sepal Length vs Petal Length")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='species', y='sepal_length', data=df)
plt.title("Sepal Length par espèce")
plt.show()
```

plt.figure(figsize=(6,4))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Matrice de corrélation")
plt.show()


Les visualisations montrent une forte corrélation entre certaines variables,
notamment entre la longueur des pétales et celle des sépales.


```python
fig = px.scatter(
    df,
    x="sepal_length",
    y="petal_length",
    color="species",
    title="Scatter interactif avec Plotly"
)
fig.show()
```

```python
# Charger un deuxième dataset : Titanic
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df2 = pd.read_csv(url)

df2.head()
```

```python
sns.countplot(x='Survived', data=df2)
plt.title("Répartition des survivants")
plt.show()
```

```python
sns.countplot(x='Sex', hue='Survived', data=df2)
plt.title("Survie par sexe")
plt.show()
```

```python
fig = px.histogram(
    df2,
    x="Age",
    color="Survived",
    title="Survie par âge (Plotly)"
)
fig.show()
```
